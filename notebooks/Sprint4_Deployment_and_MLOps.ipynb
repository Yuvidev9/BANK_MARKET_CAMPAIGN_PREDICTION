```python
import joblib
from pathlib import Path
from sklearn.pipeline import Pipeline

# Load the saved preprocessor
preprocessor = joblib.load("../models/preprocessor_sprint1.joblib")

# Load the optimized Random Forest model
final_model = joblib.load("../models/best_bank_campaign_model.pkl")

# Create the complete ML pipeline
deployment_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", final_model)
])

# Save the complete pipeline
pipeline_path = Path("../models/deployment_pipeline.joblib")

joblib.dump(deployment_pipeline, pipeline_path)

print("Complete ML pipeline created successfully!")
print("Pipeline saved at:", pipeline_path)
```

```python
import joblib
import pandas as pd

# Load complete deployment pipeline
deployment_pipeline = joblib.load(
    "../models/deployment_pipeline.joblib"
)

# Correct customer details in original column order
sample_customer = [[
    58, "management", "married", "tertiary",
    "no", 2143, "yes", "no", "unknown",
    5, "may", 1, -1, 0, "unknown", 0
]]

# Get expected input column names
preprocessor = deployment_pipeline.named_steps["preprocessor"]

# Convert into DataFrame
customer_df = pd.DataFrame(
    sample_customer,
    columns=preprocessor.feature_names_in_
)

# Predict
prediction = deployment_pipeline.predict(customer_df)

# Prediction probability
probability = deployment_pipeline.predict_proba(customer_df)

print("Prediction:", prediction[0])
print("Prediction Probability:", probability[0])

if prediction[0] == 1:
    print("Customer is likely to subscribe.")
else:
    print("Customer is unlikely to subscribe.")
```
